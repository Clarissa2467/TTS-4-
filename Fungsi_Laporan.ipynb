```python
# ==========================================
# FUNGSI BAGIAN ANGGOTA A: Input Data Penerimaan
# ==========================================
def input_data_agroindustri():
    daftar_penerimaan = []
    print("==================================================")
    print("    INPUT DATA PENERIMAAN BAHAN BAKU AGROINDUSTRI   ")
    print("==================================================")

    # Meminta jumlah transaksi/lot yang akan diinput
    while True:
        try:
            jumlah_batch = int(input("Masukkan jumlah lot/batch penerimaan: "))
            if jumlah_batch > 0:
                break
            print("Jumlah batch harus lebih dari 0!")
        except ValueError:
            print("Input tidak valid! Harap masukkan angka bulat.")

    # Loop penginputan data per batch
    for i in range(jumlah_batch):
        print(f"\n--- Transaksi Batch Ke-{i+1} ---")
        pemasok = input("Nama Pemasok / Kelompok Tani : ").strip()
        komoditas = input("Jenis Komoditas (Kopi/Sawit/Susu) : ").strip().title()

        # Validasi input berat/volume bahan baku
        while True:
            try:
                jumlah_bahan = float(input("Jumlah Bahan Baku (kg/Liter)  : "))
                harga_per_unit = float(input("Harga Beli Per Unit (Rp)      : "))
                if jumlah_bahan > 0 and harga_per_unit > 0:
                    break
                print("Jumlah dan harga harus lebih besar dari 0!")
            except ValueError:
                print("Input tidak valid! Harap masukkan angka yang sesuai.")

        # Menyimpan data dalam bentuk dictionary
        batch_data = {
            "id_batch": f"AGRO-{i+1:03d}",
            "pemasok": pemasok,
            "komoditas": komoditas,
            "jumlah_bahan": jumlah_bahan,
            "harga_per_unit": harga_per_unit,
            "total_biaya": jumlah_bahan * harga_per_unit
        }

        daftar_penerimaan.append(batch_data)
        print(f"[SUCCESS] Batch {batch_data['id_batch']} berhasil dicatat!")

    print("\n==================================================")
    print("[INFO] Seluruh data penerimaan berhasil disimpan.")
    return daftar_penerimaan


# ==========================================
# FUNGSI BAGIAN ANDA (ANGGOTA B): Laporan Rekapitulasi
# ==========================================
def buat_laporan_agroindustri(data_penerimaan):
    if not data_penerimaan:
        print("[WARNING] Belum ada data penerimaan untuk dilaporkan.")
        return

    print("\n==================================================")
    print("        LAPORAN REKAPITULASI PENERIMAAN KODE        ")
    print("==================================================")

    total_keseluruhan_biaya = 0
    total_volume_bahan = 0

    for item in data_penerimaan:
        print(f"ID Batch      : {item['id_batch']}")
        print(f"Pemasok       : {item['pemasok']}")
        print(f"Komoditas     : {item['komoditas']}")
        print(f"Jumlah        : {item['jumlah_bahan']:,.1f}")
        print(f"Harga/Unit    : Rp {item['harga_per_unit']:,.0f}")
        print(f"Total Biaya   : Rp {item['total_biaya']:,.0f}")
        print("-" * 50)

        total_keseluruhan_biaya += item['total_biaya']
        total_volume_bahan += item['jumlah_bahan']

    print(f"\n[RINGKASAN AKHIR]")
    print(f"Total Seluruh Volume/Berat Bahan Baku : {total_volume_bahan:,.1f}")
    print(f"Total Pengeluaran Pembelian         : Rp {total_keseluruhan_biaya:,.0f}")
    print("==================================================")


# ==========================================
# PROGRAM UTAMA (Integrasi Keseluruhan Sistem)
# ==========================================
if __name__ == "__main__":
    # 1. Menjalankan fungsi input data (Bagian Anggota A)
    data_bahan_baku = input_data_agroindustri()

    # 2. Menjalankan fungsi laporan rekapitulasi (Bagian Anggota B)
    buat_laporan_agroindustri(data_bahan_baku)
```
