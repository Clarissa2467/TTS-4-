```python
# ==========================================
# FUNGSI BAGIAN ANDA (ANGGOTA B): Laporan Rekapitulasi
# ==========================================
def buat_laporan_agroindustri(data_penerimaan):
    if not data_penerimaan:
        print("[WARNING] Belum ada data penerimaan untuk dilaporkan.")
        return

    print("\n==================================================")
    print("        LAPORAN REKAPITULASI PENERIMAAN KODE        ")
    print("==================================================")

    total_keseluruhan_biaya = 0
    total_volume_bahan = 0

    for item in data_penerimaan:
        print(f"ID Batch      : {item['id_batch']}")
        print(f"Pemasok       : {item['pemasok']}")
        print(f"Komoditas     : {item['komoditas']}")
        print(f"Jumlah        : {item['jumlah_bahan']:,.1f}")
        print(f"Harga/Unit    : Rp {item['harga_per_unit']:,.0f}")
        print(f"Total Biaya   : Rp {item['total_biaya']:,.0f}")
        print("-" * 50)

        total_keseluruhan_biaya += item['total_biaya']
        total_volume_bahan += item['jumlah_bahan']

    print(f"\n[RINGKASAN AKHIR]")
    print(f"Total Seluruh Volume/Berat Bahan Baku : {total_volume_bahan:,.1f}")
    print(f"Total Pengeluaran Pembelian         : Rp {total_keseluruhan_biaya:,.0f}")
    print("==================================================")
```
